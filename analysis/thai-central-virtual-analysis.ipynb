```python
data_dir = "../data/converted/thai-central-to-virtual-vctk"

data_type = "txt"

speakers = []
files = []
old_speakers = []
old_files = []
for speaker in os.listdir(os.path.join(data_dir, data_type)):
    if speaker[0] != "v":
        old_speakers.append(speaker)
        old_files.extend([file for file in os.listdir(os.path.join(data_dir, data_type, speaker))])
        continue
    speakers.append(speaker)
    files.extend([file for file in os.listdir(os.path.join(data_dir, data_type, speaker))])

print("Number of speakers:", len(speakers))
print("Number of files:", len(files))
print("Number of old speakers:", len(old_speakers))
print("Number of old files:", len(old_files))
```

```python
import os
import glob
import soundfile as sf
from tqdm import tqdm

# Path pattern
path_pattern = "../data/converted/thai-central-to-virtual-vctk/wav16_silence_trimmed/**/*.flac"

# Extract base directory dynamically from the path pattern
# Remove the '**/*.flac' part to get the base directory
base_dir = path_pattern.split("/**/")[0]
base_dir = os.path.normpath(base_dir)

# Get all flac files matching the pattern
flac_files = glob.glob(path_pattern, recursive=True)

if not flac_files:
    print(f"No FLAC files found matching pattern: {path_pattern}")
else:
    print(f"Found {len(flac_files)} FLAC files. Processing...")
    print(f"Base directory: {base_dir}")
    
    # Calculate total duration
    total_duration_seconds = 0
    
    # Track speaker folders
    speaker_folders = set()
    
    # Use tqdm for progress bar
    for flac_file in tqdm(flac_files):
        try:
            # Get audio info
            info = sf.info(flac_file)
            total_duration_seconds += info.duration
            
            # Extract speaker folder - take the directory right after the base_dir
            rel_path = os.path.relpath(os.path.dirname(flac_file), base_dir)
            if '/' in rel_path:
                speaker = rel_path.split('/')[0]  # First directory is the speaker
            else:
                speaker = rel_path  # If there's no further nesting
                
            speaker_folders.add(speaker)
            
        except Exception as e:
            print(f"Error processing {flac_file}: {e}")
    
    # Convert to hours
    total_duration_hours = total_duration_seconds / 3600
    
    # Print results
    print(f"\nTotal duration: {total_duration_hours:.2f} hours")
    print(f"                ({total_duration_hours*60:.2f} minutes)")
    print(f"                ({total_duration_hours*3600:.2f} seconds)")
    
    # Print only the number of speakers
    print(f"Number of speakers: {len(speaker_folders)}")
```
