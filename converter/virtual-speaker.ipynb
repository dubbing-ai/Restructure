```python
import os
import json
import shutil
```

```python
DEST_DIR = "../data/results"
VOICE_ID = 'cv131'
SRC_PATH = os.path.join('../data/raw/voice-dataset', VOICE_ID)

# Clean and create directories
if os.path.exists(DEST_DIR):
   print("Clearing destination folder")
   shutil.rmtree(DEST_DIR)


with open("speaker-number.json", "r") as file:
    data = json.load(file)
for k in data.keys():
    DEST_VOICE_PATH = os.path.join(DEST_DIR, k)
    os.makedirs(DEST_VOICE_PATH, exist_ok=True)
    for v in data[k]:
         SRC_VOICE_PATH = os.path.join(SRC_PATH, v)
         shutil.copy2(SRC_VOICE_PATH, DEST_VOICE_PATH)
print("Creating voice directory")
```
